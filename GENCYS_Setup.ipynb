# GENCYS 2.0 — Workshop Setup

This is the single entry point for the workshop. It downloads an independent copy of the complete public repository into this Colab runtime. No GitHub account is needed.

## One required step

Run the cell below. When it finishes, every notebook, dataset, image, and support file in the repository will be available under `/content/GENCYS_Quantum`.

> The `/content` folder is private to this runtime but temporary. It is not shared with other students and disappears when the runtime resets.

```python
from pathlib import Path
from tempfile import TemporaryDirectory
from urllib.parse import quote
import subprocess

from IPython.display import Markdown, display

GITHUB_OWNER = "awwmar"
GITHUB_REPOSITORY = "GENCYS-Quantum"
GITHUB_BRANCH = "main"
WORKSHOP_FOLDER = Path("/content/GENCYS_Quantum")

if GITHUB_OWNER == "YOUR_" + "GITHUB_USERNAME":
    raise RuntimeError(
        "The organizer must replace the GitHub-owner placeholder before publishing."
    )

repository_url = f"https://github.com/{GITHUB_OWNER}/{GITHUB_REPOSITORY}.git"

if WORKSHOP_FOLDER.exists():
    subprocess.run(
        ["git", "-C", str(WORKSHOP_FOLDER), "pull", "--ff-only"],
        check=True,
    )
    print(f"Updated the existing runtime copy: {WORKSHOP_FOLDER}")
else:
    with TemporaryDirectory(prefix="gencys_download_", dir="/content") as temp_dir:
        staged_folder = Path(temp_dir) / "repository"
        subprocess.run(
            [
                "git", "clone", "--quiet", "--depth", "1",
                "--branch", GITHUB_BRANCH, repository_url, str(staged_folder),
            ],
            check=True,
        )
        staged_folder.rename(WORKSHOP_FOLDER)
    print(f"Downloaded an independent runtime copy to: {WORKSHOP_FOLDER}")

exercise_notebooks = sorted((WORKSHOP_FOLDER / "notebooks").glob("*.ipynb"))
solution_notebooks = sorted((WORKSHOP_FOLDER / "solutions").glob("*.ipynb"))

def notebook_links(notebooks):
    links = []
    for notebook in notebooks:
        relative_path = notebook.relative_to(WORKSHOP_FOLDER).as_posix()
        colab_url = (
            f"https://colab.research.google.com/github/{GITHUB_OWNER}/"
            f"{GITHUB_REPOSITORY}/blob/{GITHUB_BRANCH}/{quote(relative_path, safe='/')}"
        )
        title = notebook.stem.replace("_", " ")
        links.append(f"- [{title}]({colab_url})")
    return "\n".join(links)

if not exercise_notebooks:
    display(Markdown("### Setup complete\nNo exercise notebooks were found yet."))
else:
    display(
        Markdown(
            "### Setup complete\n"
            f"All repository files are in `{WORKSHOP_FOLDER}`.\n\n"
            "### Exercises\n" + notebook_links(exercise_notebooks) +
            "\n\n### Solutions (open after attempting the exercise)\n" +
            notebook_links(solution_notebooks) +
            "\n\nAfter opening an exercise, choose **File → Save a copy in Drive** "
            "to keep your own editable notebook."
        )
    )
```

## Important

Opening an exercise link creates a personal notebook session; it does not change the GitHub master. If a future exercise uses data or assets, its first setup cell should download those files into that exercise's runtime.
